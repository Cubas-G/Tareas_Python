```python
## Ejercicio 1: Longitud de una cadena

""" En este ejercicio calculamos cuántos caracteres tiene una cadena.  
Usamos la función `len()`, que devuelve la longitud total incluyendo espacios. """

# Ejercicio 1: Longitud de una cadena

mensaje = "Hola, bienvenidos al curso de Python"

# len() devuelve el número total de caracteres
longitud = len(mensaje)

print("Longitud de la cadena:", longitud)
```

```python
## Ejercicio 2: Índices de cadenas
""" 
Las cadenas permiten acceder a sus caracteres mediante índices:
- El índice 0 es el primer carácter.
- Los índices negativos permiten acceder desde el final. """

# Ejercicio 2: Índices de cadenas

texto = "Python"

# Primer carácter (posición 0)
print("Primera letra:", texto[0])

# Último carácter usando índice negativo
print("Última letra:", texto[-1])
```

```python
## Ejercicio 3: Extraer partes de una dirección

""" Usamos `split()` para dividir la cadena por comas y luego extraemos:
- El nombre de la calle
- El número
- La ciudad

También usamos `strip()` para eliminar espacios sobrantes. """

# Ejercicio 3: Extraer partes de una dirección

direccion = "Calle Falsa, 123, Springfield"

# split() divide la cadena en partes separadas por comas
partes = direccion.split(",")

calle = partes[0]
numero = partes[1].strip()
ciudad = partes[2].strip()

print("Calle:", calle)
print("Número:", numero)
print("Ciudad:", ciudad)
```

```python
## Ejercicio 4: Identificador único con slicing

""" Creamos un identificador tomando:
- Las 3 primeras letras del nombre
- Las 3 primeras letras del apellido
- Los 2 últimos dígitos del año de nacimiento

Luego concatenamos todo. """

# Ejercicio 4: Identificador único

nombre = "Cristian"
apellido = "Rodriguez"
anio_nacimiento = "1989"

# Slicing para extraer partes
id_nombre = nombre[:3]
id_apellido = apellido[:3]
id_anio = anio_nacimiento[-2:]

identificador = id_nombre + id_apellido + id_anio

print("Identificador generado:", identificador)
```

```python
## Ejercicio 5: Inmutabilidad de cadenas
""" 
Las cadenas en Python son inmutables, lo que significa que no se pueden modificar directamente.
Intentar cambiar un carácter produce un error.

Para modificar una cadena, debemos crear una nueva usando slicing y concatenación. """

# Ejercicio 5: Inmutabilidad de cadenas

palabra = "robot"

# Intento de modificar un carácter (esto dará error)
try:
    palabra[0] = "R"
except TypeError as e:
    print("Error:", e)

# Forma correcta: crear una nueva cadena
palabra_modificada = "R" + palabra[1:]

print("Palabra modificada:", palabra_modificada)
```

```python
## Ejercicio 6: Analizador de palabras

""" En este ejercicio:
- Convertimos el texto a minúsculas para hacer una búsqueda insensible a mayúsculas.
- Contamos cuántas veces aparece la palabra exacta usando `split()`.
- Contamos cuántas veces aparece como subcadena usando `.count()`.
- Generamos una versión del texto donde la palabra objetivo aparece en MAYÚSCULAS. """

# Ejercicio 6: Analizador de palabras

texto = "Python es POTENTE. Aprender python es útil. PYTHON abre puertas."
objetivo = "Python"

# Convertimos ambos a minúsculas
texto_lower = texto.lower()
objetivo_lower = objetivo.lower()

# Contar apariciones como palabra exacta
palabras = texto_lower.split()
conteo_palabra = sum(1 for p in palabras if p.strip(".,") == objetivo_lower)

# Contar apariciones como subcadena
conteo_subcadena = texto_lower.count(objetivo_lower)

# Resaltar la palabra objetivo en mayúsculas
texto_resaltado = texto_lower.replace(objetivo_lower, objetivo.upper())

print("Coincidencias como palabra exacta:", conteo_palabra)
print("Coincidencias como subcadena:", conteo_subcadena)
print("\nTexto resaltado:\n", texto_resaltado)
```
